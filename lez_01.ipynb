```python
# linear algebra stuff

import numpy as np

x = np.array([[3.3, -1.7, 4.6],[7.4, 12.6, 22.3],[-3, 6.7, 7.9]])

from numpy import linalg as la

detx = la.det(x)
print(detx, '\n')

invx = la.inv(x)
print(invx, '\n')

# multiplication between arrays

not_unity = x*invx  # not row-column multiplication. This is a element-wise multiplication

unity = np.dot(x, invx) # dot product for nd arrays, so the row-column multiplication

#NOTE: you rarely will get zero, it depends onto the floating precision of your computer, which is like e-18 or whatever, PAY ATTENTION!
# this is called FLOATING POINT TRUNCATION
```
